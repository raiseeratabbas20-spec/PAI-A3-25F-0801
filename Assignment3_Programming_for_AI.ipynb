## Question 3: Sensor cube (18 marks)
`data/sensor_cube.npy` holds temperature readings from 6 sensors over 10 days, 24 hours per day. Axes are (day, hour, sensor). `data/sensor_meta.npz` holds `offsets` (calibration per sensor), `sensor_ids` and `days`. The logger writes -999 when a reading fails. Some other readings are physically impossible for this site (below 0 or above 60 degrees C).

**3a (3 marks)** Load both files. Print the shape, dtype and size of the cube, list the arrays stored in the npz archive and print `sensor_ids`. Report how many readings equal -999 and how many other readings are impossible. Why is -999 more dangerous than NaN if you take a mean directly?

answer>>
If you calculate the mean directly without filtering, standard functions treat -999 as a valid numerical value rather than missing data, which severely skews the mathematical mean downwards

````python
import numpy as np

cube = np.load('data/sensor_cube.npy')
meta = np.load('data/sensor_meta.npz')

print("Cube shape:", cube.shape)
print("Cube dtype:", cube.dtype)
print("Cube size:", cube.size)

print("Arrays in npz:", meta.files)
sensor_ids = meta['sensor_ids']
print("Sensor IDs:", sensor_ids)
offsets = meta['offsets']
days = meta['days']

num_minus_999 = np.sum(cube == -999)
num_impossible = np.sum((cube < 0) | (cube > 60)) - num_minus_999

print(f"Readings equal to -999: {num_minus_999}")
print(f"Other impossible readings (outside [0, 60]): {num_impossible}")
````

**3b (4 marks)** Make a float copy of the cube (not a view) where every failed or impossible reading is NaN. Prove with `np.shares_memory` that your copy is independent. Then apply calibration by adding `offsets` using broadcasting only, no loops. State the shape of the broadcast operand and why it aligns with the last axis. Write the check that the original file array is untouched.

````python

cube_clean = cube.astype(float).copy()

invalid_mask = (cube_clean == -999) | (cube_clean < 0) | (cube_clean > 60)
cube_clean[invalid_mask] = np.nan

print("Shares memory with original:", np.shares_memory(cube, cube_clean))

cube_calibrated = cube_clean + offsets

print("Offsets shape:", offsets.shape)
print("Aligned with last axis (sensors):", cube_calibrated.shape[-1] == offsets.shape[0])

print("Original cube still contains -999:", np.any(cube == -999))
````

**3c (4 marks)** Without loops compute: (i) the mean temperature of each sensor over all days and hours, (ii) the mean daily profile of each hour averaged over days and sensors (shape 24), (iii) the hottest (day, sensor) pair using daily means. NaN must not poison the result. State the axis you used for each and why.

````python

sensor_means = np.nanmean(cube_calibrated, axis=(0, 1))
print("Sensor means:", sensor_means)

hourly_profile = np.nanmean(cube_calibrated, axis=(0, 2))
print("Hourly profile shape:", hourly_profile.shape, hourly_profile)

daily_sensor_means = np.nanmean(cube_calibrated, axis=1)
hottest_flat_idx = np.nanargmax(daily_sensor_means)
hottest_day, hottest_sensor_idx = np.unravel_index(hottest_flat_idx, daily_sensor_means.shape)
print(f"Hottest (day, sensor) pair: Day {days[hottest_day]} (index {hottest_day}), Sensor {sensor_ids[hottest_sensor_idx]}")
````

**3d (4 marks)** For sensor T3 (index 2) compute the 3-hour moving average of the hourly series for day 4 (index 3) using `np.cumsum` and slicing only, not `np.convolve`. Fill NaN in that series with the mean of the series first. Report the length of the result and explain why it is not 24.

````python

t3_day4 = cube_calibrated[3, :, 2].copy()

mean_val = np.nanmean(t3_day4)
t3_day4[np.isnan(t3_day4)] = mean_val

cumsum = np.cumsum(np.insert(t3_day4, 0, 0))
window = 3
moving_avg = (cumsum[window:] - cumsum[:-window]) / window

print("Moving average result length:", len(moving_avg))
print("Moving average values:", moving_avg)
````

**3e (3 marks)** Count the hours per day in which at least one sensor is above 35 degrees C (use `any`), and the number of days in which all sensors stay below 35 for the whole day (use `all`). Using `np.unique` with counts on the hour of each day's peak (use the sensor-averaged series), state which hour is the most common peak. Interpret in two sentences: is the sensor network behaving like a daytime heating cycle?

````python

above_35 = cube_calibrated > 35
hours_per_day = np.sum(np.any(above_35, axis=2), axis=1)
print("Hours per day with at least one sensor > 35°C:", hours_per_day)

days_all_below_35 = np.sum(~np.any(above_35, axis=(1, 2)))
print("Number of days all sensors stayed below 35°C all day:", days_all_below_35)

sensor_avg_hourly = np.nanmean(cube_calibrated, axis=2)  
peak_hours = np.argmax(sensor_avg_hourly, axis=1)
unique_hours, counts = np.unique(peak_hours, return_counts=True)
most_common_peak = unique_hours[np.argmax(counts)]
print(f"Most common peak hour across days: Hour {most_common_peak}")
````
